```python
from qiskit import transpile
qc=QuantumCircuit(3,2)
qc.h([1,2])
qc.x([1,2])
qc.cx(0,1)
qc.h([1,2])
qc.measure([1,2],[0,1])
simulator=AerSimulator()
compile=transpile(qc,simulator)
result=simulator.run(compile,shots=10000).result()
counts=result.get_counts()
print(counts)
```

```python
qc = QuantumCircuit(50, 50)
qc.x(49)
qc.h(49)
qc.h(range(49))

for i in range(49):
    qc.cx(i, 49)

qc.h(range(49))
qc.draw("text")
```

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc = QuantumCircuit(50, 50)
qc.x(49)
qc.h(49)
qc.h(range(49))

for i in range(49):
    qc.cx(i, 49)

qc.h(range(49))

qc.measure(range(49), range(49))

simulator = AerSimulator()
bit = simulator.run(qc).result()
count = bit.get_counts()

print(count)
```
