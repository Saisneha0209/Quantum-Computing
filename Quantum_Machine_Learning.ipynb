```python
import matplotlib.pyplot as plt
import qiskit
import qiskit_machine_learning
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_squared_error,mean_absolute_error
from sklearn.decomposition import PCA
from sklearn.datasets import fetch_california_housing
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

housing = fetch_california_housing(as_frame = True)
df = housing.frame

print("First 5 row")
print(df.head())

print("DATASET INFO")
print(df.info())


print(df.describe())

x = housing.data
y = housing.target

x_train,x_test,y_train,y_test =train_test_split(x,y,test_size =0.2 , random_state = 42)

scaler = StandardScaler()
x_train_scale = scaler.fit_transform(x_train)
x_test_scale = scaler.transform(x_test)


dt_model = DecisionTreeRegressor()
dt_model.fit(x_train_scale,y_train)

dt_pred =dt_model.predict(x_test_scale)
dt_rmse = np.sqrt(mean_squared_error(y_test,dt_pred))
dt_r2 = r2_score(y_test,dt_pred)

print("---DECISION TREE---")
print(f"RMSE USING DECISION TREE:{dt_rmse}")
print(f"R2 USING DECISION TREE:{dt_r2}")

rf_model = RandomForestRegressor()
rf_model.fit(x_train_scale,y_train)
rf_pred = rf_model.predict(x_test_scale)
rf_rmse = np.sqrt(mean_squared_error(y_test,rf_pred))
rf_r2 = r2_score(y_test,rf_pred)

print("---Random Forest---")
print("RMSE USING RANDOM FOREST:",rf_rmse)
print("R2 SCORE USING RANDOM FOREST:",rf_r2)

model = LinearRegression()
model.fit(x_train_scale,y_train)

y_pred = model.predict(x_test_scale)
mse = mean_squared_error(y_test,y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test,y_pred)
print("--LINEAR REGRESSION---")
print(f"RMSE USING LINEAR REGRESSION: {rmse}")
print(f"R2 SCORE USING LINEAR REGRESSION: {r2}")
```

```python
#BY PCA
pca  = PCA(n_components= 5,random_state = 42)
x_train_pca = pca.fit_transform(x_train_scale)
x_test_pca = pca.transform(x_test_scale)

#DECISION TREE WITH PCA DATA
dt_model = DecisionTreeRegressor()
dt_model.fit(x_train_pca,y_train)

dt_pred =dt_model.predict(x_test_pca)
dt_rmse = np.sqrt(mean_squared_error(y_test,dt_pred))
dt_r2 = r2_score(y_test,dt_pred)

print("---DECISION TREE (PCA) ---")
print(f"RMSE USING DECISION TREE:{dt_rmse}")
print(f"R2 USING DECISION TREE:{dt_r2}")

#RANDOM FOREST WITH PCA

rf_model = RandomForestRegressor()
rf_model.fit(x_train_pca,y_train)
rf_pred = rf_model.predict(x_test_pca)
rf_rmse = np.sqrt(mean_squared_error(y_test,rf_pred))
rf_r2 = r2_score(y_test,rf_pred)

print("---Random Forest (PCA) ---")
print("RMSE USING RANDOM FOREST:",rf_rmse)
print("R2 SCORE USING RANDOM FOREST:",rf_r2)
```
